# Environment Setup & Verification

To confirm that the `deepfake_env` conda environment is correctly
configured with GPU-enabled PyTorch and all required libraries for the
Deepfake Detection project.

```python
import sys
import torch

print("=" * 50)
print("ENVIRONMENT CHECK")
print("=" * 50)

checks = {}

# 1. Python / kernel identity
checks["Correct environment"] = "deepfake_env" in sys.executable
print(f"Python executable : {sys.executable}")
print(f"Python version    : {sys.version.split()[0]}")

# 2. PyTorch + GPU
checks["PyTorch installed"] = True
print(f"PyTorch version   : {torch.__version__}")

cuda_ok = torch.cuda.is_available()
checks["CUDA available"] = cuda_ok
print(f"CUDA available    : {cuda_ok}")
if cuda_ok:
    print(f"GPU name          : {torch.cuda.get_device_name(0)}")
    print(f"CUDA version      : {torch.version.cuda}")
else:
    print("GPU name          : None (will run on CPU — much slower for training)")

print()
print("=" * 50)
print("SUMMARY")
print("=" * 50)
for check, passed in checks.items():
    status = "✅ PASS" if passed else "❌ FAIL"
    print(f"{status}  {check}")

if all(checks.values()):
    print("\nEnvironment is fully ready.")
else:
    print("\n⚠️  One or more checks failed — resolve before continuing to other notebooks.")
```
